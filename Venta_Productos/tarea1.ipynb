Requisitos técnicos mínimos
1.- Componentes obligatorios (mínimo 5)
Debes implementar al menos 5 componentes custom, por ejemplo:

Componente	Descripción
1. Header	Logo + nombre de la tienda
arriendo-casas\src\components\Navbar\HeaderLogo.jsx

2. SearchBar	Input controlado para búsqueda
arriendo-casas\src\components\Navbar\SearchBar.jsx

3. ProductCard	Muestra info del producto (nombre, precio, imagen, categoría)
- arriendo-casas\src\components\PropertyDetail\PropertyDetail.jsx
- arriendo-casas\src\components\Properties\Properties.jsx

4. ProductList	Renderiza lista usando map
- arriendo-casas\src\components\Home\Home.jsx Renderiza arriendo-casas\src\data\propiedades.js


5. Button	Botón reutilizable (primary / secondary)

6. Footer	Información básica
- arriendo-casas\src\components\Footer\Footer.jsx
Regla importante:

ProductCard debe recibir props

Al menos 1 componente debe manejar estado (useState)


2.- Simulación de datos
Usar un array de productos o un archivo products.json

Cada producto debe tener mínimo:

id

name

price

category

image

3.- Funcionalidad mínima
Renderizar listado de productos

Uso correcto de key al renderizar listas

Componentes separados en carpetas (/components)
